```python
import pandas as pd 
import numpy as np 
import matplotlib.pyplot as plt 
import seaborn as sns 
```

```python
df = pd.read_csv('minute_features.csv')
df.head()
```

```python
df.isna().sum()
```

```python
minute = pd.read_csv("minute_features.csv", parse_dates=["minute_ts"])
events = pd.read_csv("events.csv", parse_dates=["ts", "minute_ts"])

minute.head()
```

```python
FEATURES = ["n_events", "n_errors", "screen_on", "screen_off"]

minute[FEATURES].describe()
```

```python
from sklearn.ensemble import IsolationForest

X = minute[FEATURES]

model = IsolationForest(
    n_estimators=300,      # число деревьев
    contamination=0.05,    # какую долю строк считать аномалиями (5%)
    random_state=42,       # чтобы результат повторялся
)
model.fit(X)

minute["anomaly_score"] = -model.score_samples(X)   # знак меняем: больше = необычнее
minute["is_anomaly"] = model.predict(X) == -1       # -1 значит «аномалия»

minute["is_anomaly"].sum()
```

```python
cols = ["minute_ts", "n_events", "n_errors", "screen_on", "screen_off",
        "steps_delta", "anomaly_score"]

top = minute[minute["is_anomaly"]].sort_values("anomaly_score", ascending=False)
top[cols]
```

```python
minute.groupby("is_anomaly")[FEATURES + ["steps_delta"]].mean().round(2)
```

```python
t = top["minute_ts"].iloc[0]      # самая необычная минута

ev = events[events["minute_ts"] == t]
print(len(ev), "строк в логе за эту минуту")

ev["component"].value_counts()    # какие части приложения шумели
```

```python
ev[ev["is_error"]][["ts", "msg"]]   # ошибки в эту минуту, если были
```

```python
fig, ax = plt.subplots(figsize=(12, 4))

ax.plot(minute["minute_ts"], minute["n_events"], lw=1, label="n_events")
a = minute[minute["is_anomaly"]]
ax.scatter(a["minute_ts"], a["n_events"], c="red", s=30, label="аномалия")

ax.set_ylabel("n_events")
ax.legend()
plt.show()
```
